# Propósito da aula:
A computação em nuvem está se tornando, em larga escala, uma das tecnologias mais utilizadas no mundo moderno. Pensando nisso, a atividade hands-on proporcionará aos alunos a oportunidade de interagir diretamente com essa tecnologia, permitindo que explorem novas maneiras de dar vida às suas ideias. <br>Uma vez na nuvem, seus projetos são armazenados de forma mais segura e, dependendo das configurações, podem ser acessados publicamente ou mantidos privados. Isso possibilita aos alunos compartilhar seu progresso com terceiros, expandindo suas oportunidades de aprendizado e colaboração.<br>
Assim como em qualquer implementação tecnológica, é fundamental reforçar a segurança de nossa aplicação. Isso pode ser feito utilizando certificados para criptografar a comunicação entre cliente e servidor, além de autenticação via chave pública e privada para reduzir as chances de possíveis ataques.

# Gerando um par de chaves SSH para autênticação

Esse par de chaves usaremos para autenticarmos ao nosso servidor python na AWS, o par de chaves é composto por duas chaves, uma privada e uma publica. A chave pública precisaremos importar para o servidor e a chave privada ficará conosco. Cuidado ao guardar essa chave privada, ela deve ser apenas sua, não compartilhe ela com mais ninguem.<br>
Começaremos abrindo o CMD, apertando o atalho Win+R e executando o comando "cmd":

![](https://drive.google.com/uc?id=1KDdHAZT5A7KiXgWhurN5syTvwGlZRxLN)

Executaremos os seguintes comandos:<br>
- <code>color a</code>
- <code>mkdir .ssh</code>
- <code>cd .ssh</code><br>
Esses comandos vão servir para criarmos uma pasta .ssh e entrar nela.<br>
![](https://drive.google.com/uc?id=1UMWGfhvgrj6ndt52lYKdPMhfXhNBba4d)

Verifique se o caminho se encontra do mesmo jeito do da imagem, mudando apenas a pasta "usuario".

Para gerar as nossas chaves, executamos o seguinte comando: <code>ssh-keygen -t rsa -b 2048 -f tarde -P "" </code><br>
![](https://drive.google.com/uc?id=1nuURiEzRFetRYV-KMXXqdjeMD__F1_oL)


Para verificar se as chaves foram geradas corretamente, execute o comando: <code>dir</code><br>
Esse comando lista todo o conteudo da pasta atual.<br>
![](https://drive.google.com/uc?id=1o226Sleap5zA_QZDFNrcDOz6jOMnzbgi)

Abra a pasta no navegador de arquivos do windows e abra o arquivo "tarde.pub" usando o bloco de notas. Garanta que você esteja abrindo o arquivo "tarde.pub" com o ".pub" no final.<br>
![](https://drive.google.com/uc?id=1Ad8DpjK20VtSsR2p5rkIPYDLHTV5jZ28)

Copie todo o conteudo do arquivo.
<br>![](https://drive.google.com/uc?id=1XuCYyen2dcqjyLHXm5fy8SWAkHj1_nzz)

Agora, precisaremos importar essa chave dentro da AWS para posteriormente usa-la para nos conectarmos com nosso servidor.<br>
Abra a turma do AWS Academy, clique em "Módulos" e procure o módulo nomeado "Iniciar os laboratórios de aprendizagem da AWS Academy". Dessa até o final da página e aceite os termos de condições.
<br>![](https://drive.google.com/uc?id=1mGZkl9aESZ6pCLmnt0CfLm9kPGoXSBbL)

Clique em "Start Lab" e aguarde o botão "AWS" ficar verde. Quando ficar verde, clique no botão "AWS" para entrar na console da AWS:
<br>![](https://drive.google.com/uc?id=1lc37xngNlr9Dn5eZaiyaMnVP0Q2bfjEK)

Podemos fazer algumas configurações como o idioma e o tema do console clicando no botão de engrenagem do lado de "Norte da Vírginia".
<br>![](https://drive.google.com/uc?id=1u4jkX-JGW_HaLogFXUSa_n-bktdfG1LG)

Vamos em serviços, no menu lateral que irá abrir, procuramos "Computação" e clicamos no serviço chamado "EC2":
<br>![](https://drive.google.com/uc?id=1DcytmsqOkZrv7ko_mhmdy4xDHjRlJ02b)

No menu lateral, procuramos a opção "Pares de chaves" na categoria "Rede e segurança", depois abra o menu "Ações" e clique em "Importar par de chaves":
<br>![](https://drive.google.com/uc?id=19qRnSmFFeSgyxxMY5u-_cFkV7CPdi677)

````python
import requests  # Biblioteca para realizar requisições HTTP (usada para scraping de páginas web).
from bs4 import BeautifulSoup  # Biblioteca para fazer parsing de HTML e extrair dados de páginas web.
import json  # Biblioteca para manipular arquivos JSON (leitura e escrita).
from fastapi import FastAPI, HTTPException, status  # FastAPI cria a API, HTTPException e status são usados para erros HTTP.
from pydantic import BaseModel  # BaseModel da Pydantic é usado para validação de dados na API.
import hashlib  # Biblioteca usada para calcular hashes (utilizado aqui para senhas/chaves).
````

````python
funcoes = ['admin', 'cliente']  # Lista com as funções válidas do sistema: 'admin' e 'cliente'.
````

````python
class Usuario(BaseModel):  # Definindo o modelo de dados de um usuário.
    nome: str  # O nome do usuário é do tipo string.
    senha: str  # A senha do usuário também é uma string.
    funcao: str  # A função (admin ou cliente) do usuário é uma string.
    chave: str = None  # A chave é opcional (None por padrão), será gerada automaticamente se o usuário for novo.
````

````python
class Logar_Usuario(BaseModel): # Definindo o modelo de dados de um usuário.
    nome: str # O nome do usuário é do tipo string.
    senha: str # A senha do usuário também é uma string.
````

````python
app = FastAPI(docs_url="/api/docs", openapi_url="/api/openapi.json")  # Instancia a aplicação FastAPI, que gerencia as rotas e lógica da API.
````

````python
def hash_md5(texto):  # Função que gera um hash MD5 para um texto fornecido.
    md5_hash = hashlib.md5()  # Cria um objeto hash MD5.
    md5_hash.update(texto.encode('utf-8'))  # Atualiza o hash com o texto codificado em UTF-8.
    return md5_hash.hexdigest()  # Retorna o hash no formato hexadecimal.
````

````python
def verifica_funcao(funcao, chave):  # Função para verificar se a chave pertence a um usuário com a função correta.
    usuarios = carregar_usuarios()  # Carrega a lista de usuários a partir do arquivo JSON.
    for usuario in usuarios:  # Itera sobre todos os usuários.
        if usuario['chave'] == chave:  # Verifica se a chave do usuário corresponde à chave fornecida.
            if usuario['funcao'] == funcao:  # Verifica se a função do usuário corresponde à função esperada.
                return True  # Se a função é correta, retorna True (validação aprovada).
    raise HTTPException(status_code=status.HTTP_401_UNAUTHORIZED, detail="Sua função não permite completar essa ação!")  # Se a função for incorreta, lança um erro 401.
````

````python
def salvar_usuarios(usuarios):  # Função para salvar a lista de usuários no arquivo JSON.
    with open('usuarios.json','w') as file:  # Abre o arquivo 'usuarios.json' para escrita.
        json.dump(usuarios, file, indent=4)  # Salva a lista de usuários no formato JSON, com indentação de 4 espaços.
````

````python
def carregar_usuarios():  # Função para carregar a lista de usuários do arquivo JSON.
    with open('usuarios.json', "r") as file:  # Abre o arquivo 'usuarios.json' para leitura.
        return json.load(file)  # Lê o conteúdo do arquivo e retorna como uma lista de dicionários.
````

````python
def verifica_chave(chave):  # Função que verifica se a chave fornecida pertence a algum usuário cadastrado.
    usuarios = carregar_usuarios()  # Carrega a lista de usuários a partir do arquivo JSON.
    for usuario in usuarios:  # Itera sobre todos os usuários.
        if usuario['chave'] == chave:  # Verifica se a chave do usuário corresponde à chave fornecida.
            return True  # Se a chave for encontrada, retorna True.
    raise HTTPException(status_code=401, detail="Chave não reconhecida")  # Se a chave não for encontrada, lança um erro 401.
````

````python
@app.post("/api/usuario")  # Define a rota POST para cadastrar um novo usuário.
def cadastrar_usuario(data: Usuario, chave: str):  # Função que recebe os dados do usuário e a chave de quem está criando.
    verifica_chave(chave)  # Verifica se a chave fornecida é válida (pertence a algum usuário).
    verifica_funcao('admin', chave)  # Verifica se o usuário que está criando é um 'admin'.
    usuarios = carregar_usuarios()  # Carrega a lista de usuários.
    if data.funcao not in funcoes:  # Verifica se a função do novo usuário é válida (está na lista de funções).
        raise HTTPException(status_code=status.HTTP_406_NOT_ACCEPTABLE, detail="Função não corresponde a nenhuma cadastrada no sistema!")  # Se a função for inválida, lança um erro 406.
    for usuario in usuarios:  # Itera sobre os usuários já cadastrados.
        if usuario['nome'] == data.nome and usuario['funcao'] == data.funcao:  # Verifica se já existe um usuário com o mesmo nome e função.
            raise HTTPException(status_code=status.HTTP_409_CONFLICT, detail="Usuário já existe!")  # Se o usuário já existe, lança um erro 409.
    chave_usuario = hash_md5(f"{data.nome}{data.funcao}")  # Gera uma chave MD5 única para o novo usuário, baseada no nome e função.
    data.chave = chave_usuario  # Atribui a chave gerada ao campo 'chave' do novo usuário.
    usuarios.append(data.dict())  # Adiciona o novo usuário à lista de usuários.
    salvar_usuarios(usuarios)  # Salva a lista atualizada de usuários no arquivo JSON.
    return {"mensagem":"Usuario cadastrado com sucesso!", "usuarios no sistema": usuarios}  # Retorna uma mensagem de sucesso e a lista de usuários.
````

````python
@app.post("/api/usuario/logar")  # Define a rota POST para logar um usuário.
def logar(data: Logar_Usuario):  # Função que recebe o nome e senha do usuário.
    usuarios = carregar_usuarios()  # Carrega a lista de usuários.
    for usuario in usuarios:  # Itera sobre todos os usuários.
        if data.nome == usuario['nome'] and data.senha == usuario['senha']:  # Verifica se o nome e senha fornecidos correspondem a um usuário cadastrado.
            return {"messagem": f"Bem vindo {data.nome}! Você esta cadastrado na função '{usuario['funcao']}'", "chave": usuario['chave']}  # Se as credenciais são válidas, retorna uma mensagem de boas-vindas e a chave do usuário.
    return HTTPException(status_code=401, detail="Credenciais inválidas!")  # Se as credenciais forem inválidas, lança um erro 401.
````

````python
@app.get("/api/usuario")  # Define a rota GET para listar os usuários.
def listar_usuarios(chave: str, role: str = None):  # Função que recebe uma chave e, opcionalmente, uma função (role).
    verifica_chave(chave)  # Verifica se a chave fornecida é válida.
    usuarios = carregar_usuarios()  # Carrega a lista de usuários.
    if role != None:  # Se uma função for fornecida (role).
        result = []  # Cria uma lista vazia para armazenar os usuários com a função especificada.
        if role not in funcoes:  # Verifica se a função fornecida é válida.
            raise HTTPException(status_code=status.HTTP_406_NOT_ACCEPTABLE, detail="Função não existe! (funções: admin, cliente)")  # Se a função for inválida, lança um erro 406.
        for usuario in usuarios:  # Itera sobre todos os usuários.
            if usuario['funcao'] == role:  # Verifica se a função do usuário corresponde à função fornecida.
                result.append(usuario)  # Adiciona o usuário à lista de resultados.
        return result  # Retorna a lista de usuários com a função especificada.
    return usuarios  # Se nenhuma função foi fornecida, retorna todos os usuários.
````

````python
@app.get("/api/campeonato/{serie}/{times}")  # Define a rota GET para obter dados do campeonato brasileiro.
def tabela_brasileirao(serie: str = "a", times: int = None):  # Função que recebe a série (padrão "a") e o número de times a listar.
    url = f"https://www.gazetaesportiva.com/campeonatos/brasileiro-serie-{serie}/"  # Monta a URL com base na série fornecida.

    headers = {  # Cabeçalhos HTTP para simular uma requisição de um navegador real.
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"}

    response = requests.get(url, headers=headers).content  # Faz a requisição para a URL e obtém o conteúdo da página.

    soup = BeautifulSoup(response, 'html.parser')  # Faz o parsing do HTML da página com BeautifulSoup.

    table = soup.findAll('table')[0]  # Encontra a primeira tabela da página (que contém a classificação do campeonato).
    rows = table.findAll('tr')[1:]  # Obtém todas as linhas da tabela, exceto a primeira (que contém os cabeçalhos).
    data = []  # Lista para armazenar os dados da tabela.
    position = 1  # Inicializa a posição dos times.
    contador = 0  # Contador para limitar o número de times, caso o parâmetro 'times' seja fornecido.
    for row in rows:  # Itera sobre cada linha da tabela.
        if contador == times:  # Se o número de times atingiu o limite, interrompe o loop.
            break
        cells = row.findAll('td')  # Obtém todas as células da linha (que contêm os dados de cada time).
        img = cells[0].find('img')['src']  # Obtém a URL da imagem do time.
        line = {"posicao": position, "icon": img, "P": cells[1].text.strip(), "J": cells[2].text.strip(),
                "V": cells[3].text.strip(), "E": cells[4].text.strip(),
                "D": cells[5].text.strip(), "GP": cells[6].text.strip(), "GC": cells[7].text.strip(),
                "GC": cells[8].text.strip(),  # Extraí os dados das células e os armazena em um dicionário.
                "SG": cells[9].text.strip(), "%": cells[10].text.strip()}  # Extrai o saldo de gols (SG) e o percentual (%) de aproveitamento.
        position += 1  # Incrementa a posição do time.
        data.append(line)  # Adiciona o dicionário com os dados do time à lista.
        contador += 1  # Incrementa o contador de times.
    return data  # Retorna a lista de times com seus respectivos dados.
````
